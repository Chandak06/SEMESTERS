```python
from bs4 import BeautifulSoup as bs
from urllib.request import Request, urlopen

url = "ADD URL"

# Add headers to look like a normal browser
req = Request(url, headers={"User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/113.0.0.0 Safari/537.36"})

page = urlopen(req)
print(page)

# Extract HTML
html = page.read().decode("utf-8")
soup = bs(html, "html.parser")

# Print the text content
print(soup.get_text())
```

```python
print(soup.prettify()) #to format and structure parsed HTML/XML content for human readability in BeautifulSoup.
```

```python
soup.find_all("img") #returns a list of all <img> tags found in the parsed HTML
```

```python
#ReviewerName
#Example:"p", {"class":"_2NsDsF AwS1CA"}
Name = soup.findAll()
```

```python
#to get and print the reviewer name
#"p", {"class":"_2NsDsF AwS1CA"}
Name = soup.findAll()

# Extract name from p tag
Reviewer = []
for i in range(0,len(Name)):
    Reviewer.append(Name[i].get_text())

print(Reviewer)
```

```python
#to get the product ratings which are present
#"div", {"class": "XQDdHH Ga3i8K"}

ratings = soup.find_all()
product_ratings = [r.get_text(strip=True) for r in ratings]

print(product_ratings)
```

```python
#to get and print the product rating
#"div", {"class": "XQDdHH Ga3i8K"}
Product_Rating = soup.findAll()     


Rating = []
for i in range(0,len(Product_Rating)):
    Rating.append(Product_Rating[i].get_text())

print(Rating)
```

```python
#"p", {"class":"z9E0IG"}
Review_Title = soup.findAll()     
print(Review_Title)
```

```python
#"p", {"class":"z9E0IG"}
Review_Title = soup.findAll()    

# Extract review title from p tag
Review_Desc = []
for i in range(0,len(Review_Title)):
    Review_Desc.append(Review_Title[i].get_text())


Review_Desc[:] = [i.lstrip('\n').rstrip('\n') for i in Review_Desc]

print(Review_Desc)
```
